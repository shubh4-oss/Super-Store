# Calculate outliers using IQR

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[(df['Sales'] < lower_limit) | (df['Sales'] > upper_limit)]

print("Number of Sales Outliers:", len(outliers))
```

# Import the required library

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

select X and Y

```python
# Independent variable
X = df[['Sales']]

# Dependent variable
Y = df['Profit']
```

# Create the regression mode

```python
# Create model
model = LinearRegression()

# Train model
model.fit(X, Y)
```

# Find the coefficient and intercept

```python
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

# Predict Profit

```python
# Predict profit
Y_pred = model.predict(X)

print("First 10 Actual Profit Values:")
print(Y.head(10))

print("First 10 Predicted Profit Values:")
print(Y_pred[:10])
```

# Calculate R-squared

```python
r2 = r2_score(Y, Y_pred)

print("R-squared:", round(r2, 4))
```

# Check residuals

```python
residuals = Y - Y_pred

print("First 10 Residuals:")
print(residuals.head(10))

print("Mean Squared Error:", round(mean_squared_error(Y, Y_pred), 2))
```

# Let's also plot actual versus predicted values:

```python
plt.figure(figsize=(8, 5))

plt.scatter(Y_pred, residuals)

plt.axhline(y=0, color='red', linestyle='--')

plt.title('Residual Plot: Sales and Profit')
plt.xlabel('Predicted Profit')
plt.ylabel('Residuals')

plt.show()
```

# Prepare the data

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

pca_data = df[['Sales', 'Quantity', 'Discount', 'Profit']].dropna()

print("Shape of PCA data:", pca_data.shape)
```

# Standardise the data

```python
scaler = StandardScaler()

scaled_data = scaler.fit_transform(pca_data)

print(scaled_data[:5])
```

Apply PCA

```python
pca = PCA()

pca_result = pca.fit_transform(scaled_data)

print("Variance explained by each component:")
print(pca.explained_variance_ratio_.round(4))

print("Cumulative variance explained:")
print(pca.explained_variance_ratio_.cumsum().round(4))
```

# Check the variable loadings

```python
loadings = pd.DataFrame(
    pca.components_.T,
    columns=['PC1', 'PC2', 'PC3', 'PC4'],
    index=pca_data.columns
)

print(loadings.round(3))
```

# Create the cumulative variance graph

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(pca.explained_variance_ratio_) + 1),
    pca.explained_variance_ratio_.cumsum(),
    marker='o'
)

plt.xlabel('Number of Components')
plt.ylabel('Cumulative Variance Explained')
plt.title('Cumulative Variance Explained by PCA')
plt.grid(True)
plt.show()
```
